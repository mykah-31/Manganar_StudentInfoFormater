```python
#VERSION 2


# Define the functions
def fname():
    fn = str(input("First name: "))
    return fn.upper()

def lname():
    ln = str(input("Last name: "))
    return ln.upper()

def section():
    sec = str(input("Section: "))
    return sec.upper()

def sub():
    subj = str(input("Favorite subject: "))
    return subj.upper()

def charcount(name):
    count = len(name)
    return count

# Store the function results in variables
first = fname()
last = lname()
sec = section()
fav_sub = sub()

# Print the final output
print("\nFormatted output: ")
print("Name:", first, last)
print("Section:", sec)
print("Favorite subject:", fav_sub)
print("Character count:", charcount(first) + charcount(last))
```
